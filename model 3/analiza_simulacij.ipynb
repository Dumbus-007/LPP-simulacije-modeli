## Najpogostejši kraji srečanja z naključnima začetnima postajališčema

```python
import pandas as pd

INPUT_CSV = "random_start_sim_rez.csv"
df = pd.read_csv(INPUT_CSV)

uspesna_srecanja = df[df["srecanje_id"] != "N/A"]

pogostost_srecanj = (
    uspesna_srecanja.groupby(["srecanje_id", "srecanje_ime"])
    .size()
    .reset_index(name="count")
    .sort_values(by="count", ascending=False)
)

# Izpis rezultatov
print(pogostost_srecanj.to_string(index=False))
```
