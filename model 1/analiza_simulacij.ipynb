## Najpogostejši kraji srečanja z naključnima začetnima postajališčema

```python
import pandas as pd

INPUT_CSV = "random_start_sim_rez.csv"
df = pd.read_csv(INPUT_CSV)

uspesna_srecanja = df[df["srecanje_id"] != "N/A"]

pogostost_srecanj = (
    uspesna_srecanja.groupby(["srecanje_id", "srecanje_ime"])
    .size()
    .reset_index(name="count")
    .sort_values(by="count", ascending=False)
)

# Izpis rezultatov
print(pogostost_srecanj.to_string(index=False))
```

## Test možnih krajev srečanja

```python
import networkx as nx
df1 = pd.read_csv('random_start_sim_rez.csv')
G = nx.read_graphml('model1_frekvenca.graphml')
```

```python
nodes_to_remove = []
    
for node in G.nodes():
    # Preverimo vhodno stopnjo (in-degree)
    in_degree = G.in_degree(node)
    out_degree = G.out_degree(node)
    
    # Vozlišče odstranimo le, če ima vhodno stopnjo natanko 1
    # Izpustimo začetna vozlišča (in_degree == 0) in ponikalnice brez izhoda (out_degree == 0),
    # če želimo ohraniti končne točke.
    if in_degree == 1 and out_degree > 0:
        nodes_to_remove.append(node)

potential_meeting_points = G.nodes() - nodes_to_remove

print(f"Število preostalih vozlišč: {len(potential_meeting_points)}")
```

```python
idji_csv = set(df1['srecanje_id'].dropna().astype(int).astype(str))
idji_mp = set(potential_meeting_points)
```

```python
manjkajoci_idji = idji_csv - idji_mp

if manjkajoci_idji:
    print(f"Najdenih je {len(manjkajoci_idji)} ID-jev v csv, ki jih ni v seznamu potencialnih krajev srečanja:\n")
    for srecanje_id in manjkajoci_idji:
        print(srecanje_id)
else:
    print("Vsi ID-ji iz csv se pojavijo tudi v seznamu potencialnih krajev srečanja")
```

```python
for id in manjkajoci_idji:
    id = float(id)
    st_korakov = df1[df1['srecanje_id']==id]
    st_korakov.groupby(["srecanje_id", "stevilo_korakov"])
    print(st_korakov.to_string(index=False))
```

Sprehajalca se srečata v vozlišču z vhodno stopnjo 1, če sta sprehod pričela v enakem vozlišču in nato storila zgolj en korak.
